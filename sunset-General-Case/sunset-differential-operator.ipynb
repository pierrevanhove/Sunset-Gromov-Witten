# Sunset — the finite operator algorithm: worked example

This worksheet runs the code in Appendix A <!--`app:code`-->, *A symbolic implementation of the operator algorithm*) of [Equal-mass sunset
integrals and genus-zero local Gromov--Witten theory of Calabi--Yau
$n$-folds](), and compares its output with Definition 8.1<!--`def:operator`-->, Theorem 8.2<!--`thm:PF`--> and Example 8.5<!--`ex:operators`-->.

**Files.** Keep both in the same folder as this notebook.
* [sunset-operator_appendix-A.py](sunset-operator_appendix-A.py) is the code exactly as printed in Appendix A.
* [sunset_differential_operator.py](sunset_differential_operator.py) contains the same algorithm, unchanged, plus routines that print the result as a differential operator, in plain text and in LaTeX.

**Conventions**
* $n$ is the number of propagators, so the operator describes $L=n-1$ loops.
* The operator is
$$P_{n-1}(y,\theta_y)=\theta_y^{\,n-1}+\sum_{j\ge1}y^j\,P_j(\theta_y),\qquad \theta_y=y\frac{d}{dy},$$
with powers of $y$ written to the left.
* $y=-t$ throughout (Remark 8.3<!--`rem:unsigned`-->). With this choice the holomorphic period is $\sum_D(-1)^DA_Dy^D$, where $A_D=\sum_{d_1+\dots+d_n=D}\binom{D}{d_1,\dots,d_n}^2$.

## 3. Theorem 8.2<!--`thm:PF`-->: the leading coefficient (8.3<!--`eq:leadP`-->)

The coefficient of $\theta_y^{n-1}$ should be $\prod_{0\le j<n/2}\bigl(1+(n-2j)^2y\bigr)$. All of its roots lie on $y<0$, so the operator has no finite singular point on $y>0$.

The paper also gives one value outside the Example (§ *The operator and its certificate* in Section 17): for $n=6$ this coefficient is $1+56y+784y^2+2304y^3=(1+36y)(1+16y)(1+4y)$.

````python
print(f"{'n':>2} {'J':>2}  {'coefficient of theta^(n-1)':42s} {'eq:leadP':>8}  singular points s = -1/y")
for n in range(2, 10):
    J, _ = operator_table(n); lead, pred = leading_symbol(n)
    print(f"{n:>2} {J:>2}  {str(lead):42s} {str(lead == pred):>8}  {sorted((n-2*j)**2 for j in range((n+1)//2))}")
print("\nn=6 statement in the text, 1+56y+784y^2+2304y^3 :", leading_symbol(6)[0] == [1, 56, 784, 2304])
````

## 4. Consistency with the sunset numbers

This is an independent sanity check, not part of the proof. The paper's operator is derived by the finite construction and is not fitted to a series. The check confirms that $P_{n-1}$ annihilates the holomorphic period $\sum_D(-1)^DA_Dy^D$ through $y^{40}$, i.e. that the coefficients satisfy the recurrence $\sum_jP_j(D-j)\,a_{D-j}=0$. The sign $(-1)^D$ comes from $y=-t$.

````python
for n in range(2, 10):
    print(f"n={n}:  A_D = {sunset_numbers(n, 6)}   annihilated to y^40: {check_recurrence(n)}")
````

## 5. Outside the standing hypothesis: $n=2$ and $n=3$

The paper assumes $n\ge4$ throughout (Definition 8.1<!--`def:operator`-->; the two-loop case $n=3$ is explicitly left to [Bloch–Kerr–Vanhove](https://arxiv.org/abs/1601.08181)). The finite algorithm still runs for smaller $n$ and gives the classical operators:
* For $n=2$ (one loop), $A_D=\binom{2D}{D}$ and $P_1=\theta_y+y(4\theta_y+2)$.
* For $n=3$ (two loops), $P_2=\theta_y^2+y(10\theta_y^2+10\theta_y+3)+9y^2(\theta_y+1)^2$.

````python
from math import comb
report_operator(2)
print("  A_D == binomial(2D, D):", sunset_numbers(2, 20) == [comb(2*D, D) for D in range(21)])
display(Math(latex_operator(2)))
report_operator(3)
display(Math(latex_operator(3)))
````

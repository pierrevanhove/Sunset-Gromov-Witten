# Generic algebraic properties of the one-mass sunset multiloop integral
This worksheet reports on the generic algebraic properties of the one-mass sunset multiloop integral. 
<br>
Uniform sunset family:  
$F_L = (1,...,1)$ hypersurface in $(\mathbb P^1)^{L+1}$,  $X_L = \textrm{Tot}(K_{F_L})$.


- $L=2$ : $F_2$ = sextic del Pezzo = $\textrm{Bl}_3 \mathbb P^2$   (toric)      $X_2$ local CY 3-fold
-  $L=3$ : $F_3$ = Mori-Mukai 4-1 threefold      (NOT toric)  $X_3$ local CY 4-fold
-  $L=4$ : $F_4$ = Fano fourfold in $(\mathbb P^1)^5$      (NOT toric)  $X_4$ local CY 5-fold


Nothing is hardcoded per loop order $L$: the Picard-Fuchs operator $P_y = \sum_j y^j P_j(\theta)$ is
recovered by exact linear algebra from the sunset numbers, so the script extends to any $L$.

Some notation:

- $\gamma_L = H^{L-1}/(L-1)! = e_{L-1}(p_1,...,p_{L+1})$   (integral for every $L$)
- $F_\gamma = -((L+1)!/(L-1)!) K$,  $K = z^{-2}$ coefficient of $e^{-HT/z} I^0_X$
- Identity:  $f_{L-1} = (-1)^L \theta_Q F_\gamma$ where $Q$ is the Novikov variable.
